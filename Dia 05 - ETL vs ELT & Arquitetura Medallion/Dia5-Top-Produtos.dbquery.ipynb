```sql
----------------------------------------------
CREATE OR REPLACE TABLE `capgemini_academy`.`gold`.`top_produtos` AS
    SELECT
        produto_id,
        SUM(quantidade) AS qtde_vendida,
        SUM(valor_total) AS receita
    FROM `capgemini_academy`.`silver`.`pedidos_autoloader_v2`
    GROUP BY produto_id;
----------------------------------------------
SELECT *
FROM `capgemini_academy`.`gold`.`top_produtos`
ORDER BY receita DESC;
```
